```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({'font.size': 14})

procs = np.array([1, 2, 8, 24, 48, 64])

# Serial baselines from mm_scaling_9127569.out
serial = {200: 0.027471, 400: 0.274120, 800: 2.180941, 1000: 3.959589}

# Parallel times from mm_scaling_9127547.out
times = {
    200:  [0.027170, 0.013658, 0.020202, 0.002372, 0.005797, 0.007400],
    400:  [0.263480, 0.132256, 0.050241, 0.013523, 0.016655, 0.007670],
    800:  [2.070354, 1.038030, 0.314671, 0.139952, 0.072257, 0.083656],
    1000: [3.816651, 1.899186, 0.651854, 0.272020, 0.173405, 0.095719],
}

fig, ax = plt.subplots(figsize=(10, 6))

for N in [200, 400, 800, 1000]:
    speedups = serial[N] / np.array(times[N])
    ax.plot(procs, speedups, 'o-', label=f'N={N}')

ax.plot(procs, procs, 'k--', alpha=0.4, label='Ideal')

ax.set_xlabel('Number of processes')
ax.set_ylabel('Speedup')
ax.legend()
ax.set_xticks(procs)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('speedup.pdf', bbox_inches='tight')
plt.show()
```
